```python
import pandas as pd
```

```python
df_laptop = pd.read_csv("laptop_price.csv",encoding_errors  = "ignore")
```

```python
df_laptop.head()
```

## SYNTAX
### DataFrame.duplicated(
### subset=None,
### keep='first'
### )

```python
# find duplicate in colum 1
df_laptop[df_laptop.duplicated("laptop_ID")]
```

```python
#duplicates in more than 2 columns
duplicated = df_laptop.duplicated(["Product" , "TypeName" , "Inches"])
```

```python
df_laptop[duplicated].sort_values(["Product" , "TypeName"])
```

```python
# sort the dataframe in ascending order by company and price
df_laptop = df_laptop.sort_values(["Company" , "Price_euros"])
```

```python
duplicated_first = df_laptop.duplicated("Company",keep ="first") # keep = "first" Mark all duplicates except the first occurrence as True
df_laptop[duplicated_first]
```

```python
df_laptop[~duplicated_first][["Company","Price_euros"]]
```

```python
duplicate_keep_false = df_laptop.duplicated("Company" , keep = False) # Mark all occurrences of duplicates as True
df_laptop[duplicate_keep_false]
```
