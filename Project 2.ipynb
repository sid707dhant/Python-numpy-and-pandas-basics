```python
import pandas as pd
```

```python
df_population_raw = pd.read_csv("population_total.csv")
df_population_raw
```

# Making Pivot Table

```python
# dropping null values
df_population_raw.dropna(inplace = True)
```

```python
df_pivot = df_population_raw.pivot_table(index = "year" , columns= "country" , values = "population")
```

```python
df_pivot = df_pivot[["United States" , "India" , "China"] ]
```

# Plotting dataframes

```python
df_pivot.plot(kind = 'line' , xlabel = "Year" ,ylabel = "Population" , title = "Population ( 1955-2020 )" )
```

# Single Barplot

```python
# selecting only one year
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
df_pivot_2020
```

```python
# we have to transpose the dataframe to gave countries as index
df_pivot_2020 = df_pivot_2020.T # transpose
df_pivot_2020.plot(kind = "bar"  , xlabel= "Country (Year- 2020)" , ylabel = "Population" , title = "Population 2020")
```

# Barplot grouped by 'n' variables

```python
df_pivot_sample = df_pivot[df_pivot.index.isin([1980 , 1990 , 2000 , 2010 , 2020])]
df_pivot_sample.plot(kind= 'bar' )
```

# Pie chart



```python
df_pivot_2020.rename(columns={2020:'2020'} ,inplace = 'True')
df_pivot_2020.plot(kind= 'pie' , y = "2020")
```

# Single Boxplot

```python
df_pivot['United States'].plot(kind= 'box' ,color = 'violet' ,ylabel = 'population')
```

# Multiple Boxplots

```python
df_pivot.plot(kind = 'box' , color ='green' , ylabel = 'Population')
```

# Histogram

```python
df_pivot.plot(kind= 'hist')
```

# Scatter Plot

```python
df_sample = df_population_raw[df_population_raw['country'].isin(['United States' , 'India' , 'China' , 'Brazil' , 'Indonesia'])]
df_sample.plot(kind= 'scatter' , x= 'year' , y= 'population')
```
