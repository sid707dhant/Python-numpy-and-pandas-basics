# set_index() and sort_index method

```python
df_exam = pd.read_csv('StudentsPerformance.csv')
df_exam
```

```python
import random
```

```python
new_index = np.arange(0, 1000)
# [0, 1, 2, 3, ..., 999]

# Step 2: Shuffle them randomly
random.shuffle(new_index)
# [473, 21, 867, 5, ...]  
```

## SET INDEX

```python
# SET new_index column as index

# Step 3: Add as column
df_exam['new_index'] = new_index

# Step 4: Make it the index
df_exam.set_index('new_index', inplace=True) # inplace = True means that we want to modify the original dataframe instead of creating a new one

# Step 5: Display
df_exam
```

## SORT INDEX METHOD


```python
df_exam.sort_index(inplace=True) # ascending order
df_exam
```

```python
# sort descending
df_exam.sort_index(ascending=False)
```
