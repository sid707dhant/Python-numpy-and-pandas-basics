```python
import pandas as pd
import numpy as np
```

```python
df_laptop = pd.read_csv("laptop_price.csv",encoding_errors='ignore')
```

```python
# Syntax
# np.where(condition, value_if_true, value_if_false)
```

```python
np.where(df_laptop["Price_euros"] > 2000 , "Expensive" , "Cheap")
```

```python
# Create a new column "Price_tier"
df_laptop["Price_tier"] = np.where(df_laptop["Price_euros"] > 2000 , "Expensive" , "Cheap")
df_laptop["Price_tier"].head()
df_laptop.head()
df_laptop.value_counts("Price_tier")
```

```python
# Create an array based on screen size
df_laptop["Screen_size"] = np.where(df_laptop["Inches"]  > 15 , "Big" , "Small")
df_laptop["Screen_size"].head()
df_laptop.value_counts("Screen_size")
```

# Creating a conditional column from more than 2 choices using np.select()

```python
conditions = [ df_laptop["Price_euros"] > 3000,
            (df_laptop["Price_euros"] <= 3000  ) & (df_laptop["Price_euros"] > 2000),
            (df_laptop["Price_euros"] <= 2000  ) & (df_laptop["Price_euros"] > 800),
            (df_laptop["Price_euros"] <= 800  )

]
values = ["Too Expensive" , "Expensive" , "Affordable" , "Cheap"]
df_laptop["Price_tiers"] = np.select(conditions , values ,default="Unknown" )
df_laptop
```

```python
# Create an array based on multiple screen size tiers
# Too big >16 , big > 14 , Small > 12, Too Small <10
conditions = [ df_laptop["Inches"] >16,
              (df_laptop["Inches"] <= 16  ) & (df_laptop["Inches"] > 14),
              (df_laptop["Inches"] <= 14  ) & (df_laptop["Inches"] > 12),
              (df_laptop["Inches"] <= 10  ) 

]
values = ["Too Big" , "Big" , "Small" , "Too Small"]
df_laptop ["Screen_tier"] =  np.select(conditions , values, default= "Unknown")
df_laptop
```
