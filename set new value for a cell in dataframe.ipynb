```python
import pandas as pd
```

```python
df = pd.read_csv("players_20.csv")
df.set_index("short_name" , inplace = True)
df = df[[ "long_name" , "age" ,"dob" , "height_cm" , "weight_kg" , "nationality" , "club"]]
df
```

```python
# update l .Messi height
df.loc['L. Messi' , 'height_cm' ] = 175
```

```python
# set height of all players to 190
df.loc[:,'height_cm']=190
df
```

```python
# get all columns of all last player ranked in fifa
df.iloc[-1,:] #negative indexing
```

```python
# set null values to all columns to last player
import numpy as np
df.iloc[-1,:] = np.nan
df
```

```python
# set height of messi and cr to 175
df.loc[['L. Messi' , 'Cristiano Ronaldo'] , ['height_cm']] = 175
df
```

```python
# set values for rows matching a condition
cond = df['height_cm']>180
df.loc[cond , 'age' : 'weight_kg'] = 0
df
```

# Syntax: df.drop(labels, axis=0/1)

```python
# drop a row
# axis = 0 means rows and 1 is for column
df.drop('L. Messi' , axis = 0)

# using index parameter
df.drop(index = ['L. Messi'])
```

```python
# drop messi and ronalso and update data
df.drop(['L. Messi' , 'Cristiano Ronaldo'], axis= 0, inplace =True)
```

```python
df
```

```python
# drop column
df.drop('long_name' , axis= 1)
```

```python
# drop columns using columns synatax
df.drop(columns=['long_name' , 'age'])
```

```python
# drop last column by positio/index

df.drop(df.columns[-1] , axis = 1)
```

```python
# drop two or more columns and update data
df.drop(columns=['long_name', 'age'] )
# OR
df.drop(['long_name' , 'age'] , axis= 1 ,inplace =True )
```
