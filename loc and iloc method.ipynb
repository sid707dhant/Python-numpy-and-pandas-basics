# loc() method
## Label-based
## Inclusive
# SYNTAX = df.loc[row_label, column_label]

```python
import pandas as pd
```

```python
df = pd.read_csv("players_20.csv")
df.set_index("short_name" , inplace = True)
df = df[[ "long_name" , "age" ,"dob" , "height_cm" , "weight_kg" , "nationality" , "club"]]
df
```

```python
# get all data of lionel messi
df.loc["L. Messi" ]
```

```python
# get weight of Cristiano Ronaldo
df.loc["Cristiano Ronaldo" , "weight_kg"]
```

```python
# get all rows inside "height_cm" column
df.loc[ :, "height_cm"]
```

```python
# get all data about l messi anf ronaldo
df.loc[["L. Messi" , "Cristiano Ronaldo"]]
```

```python
# get height of l messi anf ronaldo
df.loc[["L. Messi" , "Cristiano Ronaldo"] , "height_cm"]
```

```python
# get height and weight of l messi
df.loc["L. Messi" , "height_cm" : "weight_kg"]
```

```python
# get height and weight of l messi and ronaldo
df.loc["L. Messi":"Cristiano Ronaldo" , "height_cm" : "weight_kg"]
```

## Selecting Range of data with a slice
### start : stop : step
### start and stop included

```python
players = ["L. Messi" , "Cristiano Ronaldo"]
df.loc[players , "age" : "club"]
```

```python
# select top 10 players with thier age to weight
df.loc['L. Messi': 'M. Salah', 'age':'weight_kg']
```

## Selecting based on condition

```python
df.loc[df["height_cm"]  > 180  , 'age' : 'club' ].head()
```

```python
# height above 180 from argentina
con1 = df['height_cm'] > 180
con2 = df["nationality"] == "Argentina"
df.loc[con1 & con2 , :]
```

# iloc Syntax
## df.iloc[row_position, column_position]
### Uses integer positions only — starts from 0, end is exclusive.

```python
# get height of messi
df.iloc[0,3]
```

```python
# weight of cr
df.iloc[1,4]
```

```python
# get all rows inside the "height_cm"
df.iloc[: , 3]
```

```python
# get all columns that correspond to the index "L. Messi"
df.iloc[0,:]
```

```python
# get all data about L.Messi and ronaldo
df.iloc[0:2, :]
```

```python
# get height of messi and ronaldo
df.iloc[0:2 , 3]
```

```python
# get height and weight of messi
df.iloc[0,3:5]
```

```python
# get height and weight of l . messi and ronaldo
df.iloc[0:2 , 3:5]
```

```python
# get age, dob, height and weight of top 10 players
df.iloc[0:10 , 1:5]
```

```python
# select players with height above 180 cm from age to weight
cond = df['height_cm']  >180 
df.iloc[list(cond) , 1:5]
```

```python
# select players with height above 180 cm and nationality is argentina from age to nationality
cond1 = df['height_cm'] > 180
cond2 = df['nationality'] == 'Argentina'
df.iloc[list(cond1 & cond2) , 1:6]
```
